```python
df_clean = df.copy()
```

```python
df_clean["Age"] = df_clean["Age"].fillna(
    df_clean["Age"].median()
)
```

```python
df_clean["Cabin"] = df_clean["Cabin"].fillna("Unknown")
```

```python
df_clean["Embarked"] = df_clean["Embarked"].fillna(
    df_clean["Embarked"].mode()[0]
)
```

```python
df_clean.isnull().sum()
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows before removal:", duplicate_count)
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
print("Duplicate rows after removal:", df_clean.duplicated().sum())
```

```python
df_clean["Sex"] = df_clean["Sex"].str.lower()
```

```python
df_clean["Embarked"] = df_clean["Embarked"].str.upper()
```

```python
df_clean["PassengerId"] = df_clean["PassengerId"].astype(str)
```

```python
df_clean.dtypes
```

```python
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]
    
    return outliers, lower_bound, upper_bound
```

```python
numerical_columns = ["Age", "Fare", "SibSp", "Parch"]

for column in numerical_columns:
    outliers, lower, upper = detect_outliers_iqr(df_clean, column)
    
    print(f"\n{column}")
    print("Lower Bound:", round(lower, 2))
    print("Upper Bound:", round(upper, 2))
    print("Outliers:", len(outliers))
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Row Count",
        "Missing Values",
        "Duplicate Rows"
    ],
    "Before Cleaning": [
        len(df),
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    "After Cleaning": [
        len(df_clean),
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum()
    ]
})

comparison
```
